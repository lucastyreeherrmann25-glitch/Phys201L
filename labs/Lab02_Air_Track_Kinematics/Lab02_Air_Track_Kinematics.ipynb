# Introduction

Before one can study the action of forces on objects, one must be able to describe motion. This description of motion is called $kinematics$. We will be learning how to use two tools to investigate the meanings of the three kinematic equations for motion in one dimension:  
The definition of average velocity:

\begin{equation}
v_{avg} \equiv \frac{\Delta x}{\Delta t}
\tag{1}
\end{equation}

The definition of average acceleration:

\begin{equation}
a_{avg} \equiv \frac{\Delta v}{\Delta t}
\tag{2}
\end{equation}
And the position of an object undergoing constant acceleration:

\begin{equation}
x(t) = \frac{1}{2} a t^2 + v_o t + x_o
\tag{3}
\end{equation}
In this lab, your job will be to compare different measurements of velocity and acceleration for an air-cart experiencing both no acceleration and constant acceleration.

# Theory

### Measurement of Velocity and Acceleration
An acoustic sensor will measure the position of a cart on the airtrack at specific intervals of time. By use of the definition of velocity (Equation ([1](#mjx-eqn))), we can compute the average velocity between two measurements. Notice that if we plot the position versus time, the fraction is the same as the rise divided by the run-- or the slope of the graph. By plotting *x* vs. *t*, we then find the slope of the graph and determine the velocity.

This same technique works for determining acceleration. However, notice that the variables to plot are now *velocity* versus time.

### Analysis of Uncertainties
It will be important to know what the uncertainties are in our computed velocities and accelerations.

First, recall that a difference in position or time is a subtraction of two quantities:
$$ \Delta x = x_2 - x_1 \nonumber $$

Therefore, we can compute the uncertainty in the difference, assuming instrumental/uniform uncertainties:
$$ \delta(\Delta x) = \delta x_2 + \delta x_1 \nonumber $$

However, in this case, the uncertainties $\delta x_1$ and $\delta x_2$ should be the same, so we can say that the uncertainty in our computed difference is:
\begin{equation}
\delta(\Delta x) = 2 (\delta x)
\tag{4}
\end{equation}

where $\delta x$ is the uncertainty in a single position measurement. We can say the same for time differences:
\begin{equation}
\delta(\Delta t) = 2 (\delta t)
\tag{5}
\end{equation}

where $\delta t$ is the uncertainty in a single time measurement. Notice that the uncertainty in the difference is larger than the uncertainty in each individual measurement.

Computing the uncertainty in velocity, $\delta v$, is straightforward. Looking at Equation ([1](#mjx-eqn)) we have:
$$ \frac{\delta v}{v} = \frac{\delta (\Delta x)}{\Delta x} + \frac{\delta (\Delta t)}{\Delta t} \nonumber $$

This equation needs some work to become useful. Capstone will take data points at constant $\Delta t$. But because $\Delta x$ may be different for each measurement, we should try to remove it from our equation. We do this by using the definition of velocity to replace $\Delta x$ with $v \Delta t$:
$$ \frac{\delta v}{v} = \frac{\delta (\Delta x)}{v \Delta t} + \frac{\delta (\Delta t)}{\Delta t} \nonumber $$

Factoring out $\Delta t$ so that we need only use it once:
$$ \frac{\delta v}{v} = \frac{1}{\Delta t} \left(\frac{\delta (\Delta x)}{v} + \delta (\Delta t) \right) \nonumber $$

We may also multiply through by $v$:
\begin{equation}
\delta v = \frac{1}{\Delta t} \left(\delta (\Delta x) + v \delta (\Delta t) \right)
\tag{6}
\end{equation}

Now we have an equation to find our uncertainty in velocity based on things we know. Notice that the uncertainty in the velocity increases with the velocity. Also notice that the uncertainty is not zero when the velocity is zero. Does this make sense?

```python
# Setup, data-source configuration, and fitting helpers

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# ---------- TEAM CONFIGURATION ----------
# Leave True while checking the supplied example. Change to False after your
# team's CSV files have been uploaded to the Lab 2 directory in your fork.
USE_EXAMPLE_DATA = False

GITHUB_OWNER = "lucastyreeherrmann25-glitch"
REPOSITORY = "Phys201L"
BRANCH = "main"
LAB_PATH = "labs/Lab02_Air_Track_Kinematics"

SENSOR_FILE = "TeamLTH_Sensor_test.csv"
LEVEL_FILE = "TeamLTH_Level_Airtrack.csv"
INCLINED_FILE = "TeamLTH_Inclined_Airtrack.csv"

# Adjust these limits only after inspecting your team's graphs. Choose the
# interval containing the clean motion between bumper collisions.
LEVEL_FIT_MIN, LEVEL_FIT_MAX = 0.20, 2.10
INCLINED_FIT_MIN, INCLINED_FIT_MAX = 0.20, 1.20

def load_lab_csv(example_name, team_name):
    """Load bundled example data or a CSV committed to the team's GitHub fork."""
    filename = example_name if USE_EXAMPLE_DATA else team_name
    local = Path(filename)
    if local.exists():
        source = str(local)
    else:
        if not USE_EXAMPLE_DATA and "REPLACE_" in GITHUB_OWNER:
            raise ValueError("Enter the repository manager's GitHub username in GITHUB_OWNER.")
        owner = "brash99" if USE_EXAMPLE_DATA else GITHUB_OWNER
        source = f"https://raw.githubusercontent.com/{owner}/{REPOSITORY}/{BRANCH}/{LAB_PATH}/{filename}"
    print(f"Loading: {source}")
    return pd.read_csv(source)

def finite_interval(x, y, sigma, xmin, xmax):
    x, y, sigma = map(lambda v: np.asarray(v, dtype=float), (x, y, sigma))
    mask = np.isfinite(x) & np.isfinite(y) & np.isfinite(sigma)
    mask &= (x >= xmin) & (x <= xmax)
    return x[mask], y[mask], sigma[mask]

def linear_fit(x, y, sigma, xmin, xmax):
    x, y, sigma = finite_interval(x, y, sigma, xmin, xmax)
    model = lambda t, intercept, slope: intercept + slope*t
    pars, cov = curve_fit(model, x, y, sigma=sigma, absolute_sigma=True)
    errors = np.sqrt(np.diag(cov))
    grid = np.linspace(x.min(), x.max(), 300)
    plt.plot(grid, model(grid, *pars), 'r-', label='Linear fit')
    return pars[0], pars[1], errors[0], errors[1]

def quadratic_fit(x, y, sigma, xmin, xmax):
    x, y, sigma = finite_interval(x, y, sigma, xmin, xmax)
    model = lambda t, A, B, C: A*t**2 + B*t + C
    pars, cov = curve_fit(model, x, y, sigma=sigma, absolute_sigma=True)
    errors = np.sqrt(np.diag(cov))
    grid = np.linspace(x.min(), x.max(), 300)
    plt.plot(grid, model(grid, *pars), 'g-', label='Quadratic fit')
    return (*pars, *errors)

def constant_fit(x, y, sigma, xmin, xmax):
    x, y, sigma = finite_interval(x, y, sigma, xmin, xmax)
    weights = 1/sigma**2
    mean = np.sum(weights*y)/np.sum(weights)
    uncertainty = np.sqrt(1/np.sum(weights))
    plt.axhline(mean, color='r', label='Weighted mean')
    return mean, uncertainty

# Sensor-test example
sensor_df = load_lab_csv('TeamLTH_Sensor_test.csv', SENSOR_FILE)
sensor_time = sensor_df['Time (s) Run #1'].to_numpy(dtype=float)
sensor_position = sensor_df['Position, Ch1&2 (m) Run #1'].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(sensor_time, sensor_position, 'b.', markersize=3)
ax.plot(sensor_time, sensor_position, 'b--', linewidth=0.7)
ax.set(xlabel='Time (s)', ylabel='Position (m)', title='Acoustic-sensor alignment test')
ax.grid(alpha=0.25)
plt.show()
```
